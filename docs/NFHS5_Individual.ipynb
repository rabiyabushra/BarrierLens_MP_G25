```python
import pandas as pd

cols_needed = [
    # Identifiers
    'caseid', 'v001', 'v002', 'v021', 'v024', 'v025',
    
    # Woman background
    'v012', 'v013', 'v106', 'v130', 'v131',
    'v501', 'v717',
    
    # Household
    'v190', 'v169a', 'v170', 'v481',
    
    # Media exposure
    'v157', 'v158', 'v159',
    
    # Autonomy
    'v743f', 'v466',
    
    # BARRIER TARGETS - Stage 1 (include f/i if present in DTA)
    'v467b',  # permission
    'v467c',  # money
    'v467d',  # distance
    'v467e',  # transport
    'v467f',  # companion (not in India CSV export)
    'v467g',  # no female provider
    'v467h',  # no provider
    'v467i',  # no drugs (not in India CSV export)
    
    # Stage 2 health outcomes
    'v626a',  # unmet need
    'm14',    # ANC visits
    
    # Bonus barrier context
    's245a',  # couldn't afford treatment
    's245b',  # couldn't afford transport
    's245h',  # husband didn't give permission
]

# Filter only columns that exist in dataset
reader = pd.read_stata('IAIR7EFL.DTA', iterator=True)
available = list(reader.variable_labels().keys())
reader.close()

cols_to_load = [c for c in cols_needed if c in available]
print(f"Loading {len(cols_to_load)} columns...")

df = pd.read_stata('IAIR7EFL.DTA', columns=cols_to_load)
print(f"Shape: {df.shape}")
df.to_csv('NFHS5_Individual.csv', index=False)
print("Saved successfully!")
```

```python
import pandas as pd

# Correct way to peek at columns
reader = pd.read_stata('IAIR7EFL.DTA', iterator=True)
all_columns = reader.variable_labels()
print(f"Total columns: {len(all_columns)}")
# Print first 50 keys
for i, col in enumerate(list(all_columns.keys())[:50]):
    print(col, ':', all_columns[col])
reader.close()
```

```python
import pandas as pd

reader = pd.read_stata('IAIR7EFL.DTA', iterator=True)
labels = reader.variable_labels()
reader.close()

# Search for anything related to barriers or healthcare access
for col, label in labels.items():
    if any(word in label.lower() for word in 
           ['barrier', 'problem', 'permission', 
            'distance', 'transport', 'money', 
            'provider', 'medicine', 'treatment']):
        print(f"{col} : {label}")
```
